```python
import speech_recognition as sr
from gtts import gTTS
import os
from datetime import datetime
import pyjokes
import wikipedia
import webbrowser
import winshell
from pygame import mixer
import time

# Inicializa o mixer globalmente
mixer.init()

def speak(text):
    print(f"Assistente: {text}")
    filename = "voice.mp3"
    
    # Gera o áudio
    tts = gTTS(text=text, lang='en')
    tts.save(filename)
    
    # Toca o áudio usando o mixer (mais estável)
    mixer.music.load(filename)
    mixer.music.play()
    
    # ESPERA terminar de falar antes de seguir
    while mixer.music.get_busy():
        time.sleep(0.1)
    
    mixer.music.unload() # Libera o arquivo para ser deletado/sobrescrito

def get_audio():
    r = sr.Recognizer()
    with sr.Microphone() as source:
        r.pause_threshold = 0.8
        r.adjust_for_ambient_noise(source, duration=1)
        print("\nOuvindo...")
        audio = r.listen(source)
        said = ""
        try:
            said = r.recognize_google(audio, language='en-US')
            print(f"Você disse: {said}")
        except Exception:
            return "" # Retorna vazio se não entender nada
    return said.lower()

def respond(text):
    if not text: # Se o texto for vazio, ignora
        return

    if 'youtube' in text:
        speak("What should I look for?")
        keyword = get_audio()
        if keyword:
            webbrowser.open(f"https://www.youtube.com/results?search_query={keyword}")
            speak(f"Found it! Here is {keyword} on Youtube")

    elif 'search' in text:
        speak("What is the topic?")
        query = get_audio()
        if query:
            try:
                # Limitamos a 2 sentenças para não ficar cansativo
                result = wikipedia.summary(query, sentences=2)
                speak("According to Wikipedia...")
                print(result)
                speak(result)
            except wikipedia.exceptions.DisambiguationError:
                speak("There are too many results for this. Be more specific.")
            except:
                speak("I couldn't find information on that.")

    elif 'joke' in text:
        speak(pyjokes.get_joke())

    elif 'empty recycle bin' in text:
        try:
            winshell.recycle_bin().empty(confirm=False, show_progress=False, sound=True)
            speak("Done! Recycle bin is empty.")
        except:
            speak("The bin is already empty or I can't access it.")

    elif 'time' in text:
        speak(f"The time is {datetime.now().strftime('%H:%M')}")

    elif 'music' in text or 'song' in text:
        # Encontra a pasta de música do usuário atual automaticamente
        music_dir = os.path.join(os.path.expanduser("~"), "Music")
        try:
            songs = [s for s in os.listdir(music_dir) if s.endswith(".mp3")]
            if songs:
                speak("Playing your music.")
                mixer.music.load(os.path.join(music_dir, songs))
                mixer.music.play()
            else:
                speak("I found no MP3 files in your music folder.")
        except FileNotFoundError:
            speak("I couldn't find your music folder.")

    elif 'exit' in text or 'stop' in text:
        speak("Goodbye! Have a nice day.")
        exit()

# --- Bloco Principal ---
speak("System online. How can I help you?")
while True:
    command = get_audio()
    respond(command)
```
