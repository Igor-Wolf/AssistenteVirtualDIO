```python
from gtts import gTTS
from IPython.display import Audio

# --- Exemplo em Inglês ---
text_to_say = "How are you doing?"
language = "en"

gtts_object = gTTS(text=text_to_say, lang=language, slow=False)
# Alterado para .mp3 para ser fiel ao formato da biblioteca
gtts_object.save("english.mp3") 
Audio("english.mp3")

```

```python
# --- Exemplo em Francês ---
french_text = "Je vais au supermarché"
french_language = "fr"

french_gtts_object = gTTS(text=french_text, lang=french_language, slow=True)
french_gtts_object.save("french.mp3")

# Para ouvir no Colab/Jupyter (basta chamar a última linha)
Audio("french.mp3")
```

```python
# --- Exemplo em Português  ---
pt_text = "Olá como você está?"
language = "pt"

pt_gtts_object = gTTS(text=pt_text, lang=language, slow=True)
pt_gtts_object.save("portuguese.mp3")

# Para ouvir no Colab/Jupyter (basta chamar a última linha)
Audio("portuguese.mp3")
```
